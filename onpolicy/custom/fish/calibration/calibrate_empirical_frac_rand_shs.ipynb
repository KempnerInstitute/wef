## Plot the baseline values for an actual agent instance

```python
from MAEFish import EFishAgent
agent = EFishAgent(num_agents=1, 
                   arena_size=(100, 100), 
                   max_step=0, 
                   max_turn=0, 
                   agent_id=0,
                   )
agent.mormyromast_angles # in radians 
agent.morm_virtual_baselines.shape  # (5, 36) = (num_sets, num_sensors)

# Plot first set of virtual baselines vs. angles in blue
# Plot other virtual baselines vs. angles in green
plt.figure(figsize=(8, 4))
plt.scatter(agent.mormyromast_angles, agent.morm_virtual_baselines[0, :], c='b')
for i in range(1, agent.morm_virtual_baselines.shape[0]):
    plt.scatter(agent.mormyromast_angles, agent.morm_virtual_baselines[i, :], c='g')
plt.xlabel('Sensor Angle (radians)')
plt.ylabel('Baseline magnitude (mV/cm)')
# plt.yscale('log')
plt.yscale('symlog', linthresh=1e-5)  # linear within ±1e-2, log outside
plt.title('Mormyromast Virtual Baselines vs. Sensor Angles')
plt.grid()
plt.legend(['Self-EOD'] + ["Cons-EOD"])
plt.savefig('mormyromast_virtual_baselines_vs_angles.png', dpi=150)
plt.show()
```
