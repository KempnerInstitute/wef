# Seed Ranking — Homing

Lightweight ranking of homing training seeds by navigation performance.
Run this first to identify the best seeds; use `degen_homing_shs.ipynb` for deeper analysis.

```python
from pathlib import Path
import utils_seeds as us
import utils_seeds_homing as ush

# ── CONFIGURE HERE ────────────────────────────────────────────────────────────
RESULTS_DIR       = Path("/home/satsingh/kr/mfrefactor/onpolicy/custom/fish/results")
GROUP_FOLDER_NAME = "Homing20260609ConsNoise0.5dynamicT2MFood0.0"
TRAIN_FINAL_WINDOW = 5
# ─────────────────────────────────────────────────────────────────────────────

run_dirs = us.discover_run_dirs(RESULTS_DIR, GROUP_FOLDER_NAME)
print(f"Group : {GROUP_FOLDER_NAME}")
print(f"Runs  : {len(run_dirs)}")
for r in run_dirs:
    print(f"  {r.parent.name} / {r.name}")
```

```python
df = ush.load_homing_ranking_df(run_dirs, train_final_window=TRAIN_FINAL_WINDOW)
df = ush.rank_homing_seeds(df)
print(f"Loaded {len(df)} seeds.")
```

```python
import pandas as pd
from IPython.display import display

pd.set_option("display.max_columns", None)
show = [
    "seed",
    "ever_reached_percent",
    "success_percent",
    "mean_distance_closure",
    "mean_end_distance",
    "median_end_distance",
    "mean_initial_distance",
    "mean_steps_to_target_reached",
    "mean_steps_to_target_success",
    "train_final_reward",
    "train_final_r_homing",
    "combined_rank",
]
show = [c for c in show if c in df.columns]
ranked = df[show].sort_values("combined_rank").reset_index(drop=True)
float_cols = ranked.select_dtypes("float").columns.tolist()
grad_cols = [c for c in ["ever_reached_percent", "mean_distance_closure", "success_percent"] if c in ranked.columns]
display(
    ranked.style
    .background_gradient(cmap="RdYlGn", subset=grad_cols, axis=0)
    .format({c: "{:.3f}" for c in float_cols})
)
```
